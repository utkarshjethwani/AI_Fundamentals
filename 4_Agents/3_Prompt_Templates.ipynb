## Chains

```python
#step 1: Create a Promt Template with a variable {topic} in the user message.
messages = [{"role": "user", "content": "write a poem about the {topic}"}, 
            {"role": "system", "content": "You are a poetry expert"}
            ]
prompt_template1 = ChatPromptTemplate.from_messages(messages)

#step 2: Define the structured LLM with the structured output, 
# create a structured output class to define the output format of the model response.
from pydantic import BaseModel, Field
class PoemOutput(BaseModel):
    topic: str = Field(description="The topic of the poem")
    poem: str = Field(description="The poem generated by the model in response to the topic")
    total_tokens: int = Field(description="Total number of tokens used in the LLM response")

llm_structured = llm.with_structured_output(PoemOutput)

#step 3: parsing the output of the model response to the structured output class using PydanticOutputParser
from langchain_core.output_parsers import PydanticOutputParser
parser = PydanticOutputParser(pydantic_object=PoemOutput)

#final step: Put it all together, using Langchain Expression Language(LCEL)
chain1 = prompt_template1 | llm_structured # | parser
```

```python
chain1.invoke({"topic": "beauty of nature"})
```
