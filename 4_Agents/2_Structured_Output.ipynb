```python
from langchain_openai import ChatOpenAI
llm = ChatOpenAI(model_name="gpt-5-mini", temperature=0)
from dotenv import load_dotenv
load_dotenv()  # Load environment variables from .env file
```

```python
response = llm.invoke("what is the capital of France?.generate the answer in a structured format - Query: <query>, Answer: <answer>")
result = response.content
result
```

## Pydantic

```python
from pydantic import BaseModel, Field

# Create a class inheriting from BaseModel to define the structured format
class QueryAnswer(BaseModel):
    query: str = Field(description="The question asked to the model")
    answer: str = Field(description="The answer generated by the model in response to the query")
    total_tokens: int = Field(description="Total number of tokens used in the LLM response")

# having Field descriptions helps the model with documentation and understanding the purpose of each field. 
# Otherwise Field is optional and can be omitted if not needed.
```

```python
object = QueryAnswer(query="what is the capital of France?", answer="The capital of France is Paris.", total_tokens=10)
object
```

```python
#link llm model with the structured output class
llm_structured = llm.with_structured_output(QueryAnswer)
response = llm_structured.invoke("what is the capital of France?")
```
